```python
import pandas as pd
import numpy as np
```

## 1. Data Loading & Cleaning

```python
# load the csv file
df = pd.read_csv('sales_data.csv')
print("\nfirst 5 rows:")
df.head()
```

```python
# сonvert order_date to datetime
df['order_date'] = pd.to_datetime(df['order_date'])

# сreate revenue column
df['revenue'] = df['price'] * df['quantity']

# verify the changes
print("updated data types:")
print(df.dtypes)
```

## 2. Revenue Analysis (Using Pandas)

```python
# total revenue
total_revenue = df['revenue'].sum()
print(f"Total Revenue: ${total_revenue:,.2f}")
```

```python
# average order value
average_order_value = df['revenue'].mean()
print(f"Average Order Value: ${average_order_value:,.2f}")
```

```python
# top 5 products by revenue
top_5_products = df.groupby('product')['revenue'].sum().sort_values(ascending=False).head(5)
print("top 5 products by revenue:")
print(top_5_products.to_frame().reset_index())
print(f"\ntotal revenue from top 5 products: ${top_5_products.sum():,.2f}")
```

```python
# revenue by region
revenue_by_region = df.groupby('region')['revenue'].sum().sort_values(ascending=False)
print("revenue by region:")
print(revenue_by_region.to_frame().reset_index())
print(f"\nregion with highest revenue: {revenue_by_region.idxmax()} (${revenue_by_region.max():,.2f})")
```

```python
# monthly sales trend
df['month'] = df['order_date'].dt.to_period('M')

monthly_sales = df.groupby('month')['revenue'].sum()
print("monthly sales trend:")
print(monthly_sales.to_frame().reset_index())
print(f"\nbest performing month: {monthly_sales.idxmax()} with ${monthly_sales.max():,.2f}")
```

## 3. Statistical Analysis (Using NumPy)

```python
# convert revenue to numpy array for statistical analysis
revenue_array = df['revenue'].to_numpy()

# mean of order revenue
mean_revenue = np.mean(revenue_array)
print(f"mean order revenue: ${mean_revenue:,.2f}")

# median of order revenue
median_revenue = np.median(revenue_array)
print(f"median order revenue: ${median_revenue:,.2f}")

# standard deviation of order revenue
std_revenue = np.std(revenue_array)
print(f"standard deviation: ${std_revenue:,.2f}")
```

```python
# BONUS: Identify Outliers using IQR Method
Q1 = np.percentile(revenue_array, 25)
Q3 = np.percentile(revenue_array, 75)
IQR = Q3 - Q1

# Define outlier bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Q1 (25th percentile): ${Q1:,.2f}")
print(f"Q3 (75th percentile): ${Q3:,.2f}")
print(f"IQR: ${IQR:,.2f}")
print(f"\noutlier bounds:")
print(f"lower bound: ${lower_bound:,.2f}")
print(f"upper bound: ${upper_bound:,.2f}")

# Find outliers
outliers = revenue_array[(revenue_array < lower_bound) | (revenue_array > upper_bound)]
print(f"\nnumber of outliers: {len(outliers)}")
if len(outliers) > 0:
    print(f"outlier values: {outliers}")
```

## 4. Customer Analysis

```python
# number of unique customers
unique_customers = df['customer_id'].nunique()
print(f"number of unique customers: {unique_customers}")
```

```python
# top 10 customers by total spending
customer_spending = df.groupby('customer_id')['revenue'].sum().sort_values(ascending=False)
top_10_customers = customer_spending.head(10)

print("top 10 customers by total spending:")
print(top_10_customers.to_frame().reset_index())
print(f"\ntop customer: {top_10_customers.idxmax()} with ${top_10_customers.max():,.2f} total spending")
```

```python
# percentage of revenue from top 20% customers (pareto analysis)
total_customers = len(customer_spending)
top_20_percent_count = int(np.ceil(total_customers * 0.20))

# get revenue from top 20% customers
top_20_percent_customers = customer_spending.head(top_20_percent_count)
revenue_from_top_20 = top_20_percent_customers.sum()

# Calculate percentage
percentage_from_top_20 = (revenue_from_top_20 / total_revenue) * 100

print(f"total number of customers: {total_customers}")
print(f"top 20% of customers: {top_20_percent_count} customers")
print(f"revenue from top 20% customers: ${revenue_from_top_20:,.2f}")
print(f"percentage of total revenue from top 20% customers: {percentage_from_top_20:.2f}%")
```
