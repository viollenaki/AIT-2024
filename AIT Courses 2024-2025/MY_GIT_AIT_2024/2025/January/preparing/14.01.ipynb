```python
def euclid_algo(a,b):
    if b == 0:
        return a
    else:
        return euclid_algo(b,a%b)
    
print(euclid_algo(36,60))
```

```python
def gcd(a,b):
    while b !=0:
        a,b = b,a%b
    return a

print(gcd(60,36))
```

```python
def lcd(a,b):
    return a*b//gcd(a,b)

print(lcd(36,60))
```

```python
import math

def next_or_equal_power_of_two(n):
    if n <= 1:
        return 0

    power = math.log2(n)

    if power.is_integer():
        return int(power)
    
    return math.ceil(power)

print(next_or_equal_power_of_two(31))
```

```python
def shift(a):
    rows = len(a)
    col = len(a[0])
    shifted = [[0 for i in range(col)] for j in range(rows)]
    for i in range(rows):
        for j in range(col):
            if i == 0 and j == 0:
                shifted[i][j] = a[rows - 1][col - 1]
            elif j == 0:
                shifted[i][j] = a[i-1][col-1]
            else:
                shifted[i][j]= a[i][j-1]

    return shifted

for i in shift([[1,2,3],[4,5,6],[7,8,9]]):
    print(i)
```

```python
def move_zeroes(nums):
    non_zero_index = 0  # Счетчик ненулевых элементов
    
    # Перемещаем все ненулевые элементы в начало массива
    for num in nums:
        if num != 0:
            nums[non_zero_index] = num
            non_zero_index += 1
    
    # Заполняем оставшуюся часть массива нулями
    for i in range(non_zero_index, len(nums)):
        nums[i] = 0

    return nums

print(move_zeroes([1,2,0,9,66,0]))
```

```python
from random import randint

# Generate a list of random integers
arr = [randint(0, 100) for i in range(100)]

# Create a dictionary to count occurrences of each number
counts = {}
for i in arr:
    counts[i] = counts.get(i, 0) + 1

# Find the mode (the most frequent number)
max_count = max(counts.values())
modes = [i for i, j in counts.items() if j == max_count]

# Print the result
if len(modes) == 1:
    print(f'The mode is: {modes[0]}')
else:
    print('There is no mode (multiple numbers have the highest frequency).')
```
