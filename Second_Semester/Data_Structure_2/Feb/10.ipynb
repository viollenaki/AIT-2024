```python
class TreeNode:
    def __init__(self, value):
        self.value = value
        self.next = None
        
class LinkedList:
    def __init__(self):
        self.head = None
        
    def check(self, value):
        current = self.head
        print(current.value)
        print(self.head.value)
        current.value = 10
        print(current.value)
        print(self.head.value)
        
linked_list = LinkedList()
linked_list.head = TreeNode(5)
linked_list.check(5)

```
