```python
def calc(nums, ops):
    b = nums.pop()
    a = nums.pop()
    op = ops.pop()

    if op == '+':
        nums.append(a + b)
    elif op == '-':
        nums.append(a - b)
    elif op == '*':
        nums.append(a * b)
    elif op == '/':
        nums.append(a / b)


def precedence(op):
    if op in ('+', '-'):
        return 1
    if op in ('*', '/'):
        return 2
    return 0


def evaluate_expression(s):
    nums = []
    ops = []
    
    if s[0] == '-':
        s = '0' + s

    i = 0
    while i < len(s):

        if s[i] == ' ':
            i += 1
            continue

        # handle number
        if s[i].isdigit():
            x = 0
            while i < len(s) and s[i].isdigit():
                x = x * 10 + int(s[i])
                i += 1
            nums.append(x)
            continue

        # handle '('
        if s[i] == '(':
            ops.append(s[i])

        # handle ')'
        elif s[i] == ')':
            while ops[-1] != '(':
                calc(nums, ops)
            ops.pop()  # remove '('

        # handle operator
        elif s[i] in "+-*/":
            # unary minus
            if s[i-1] in "(+-*/":
                nums.append(0)
            while ops and precedence(ops[-1]) >= precedence(s[i]):
                calc(nums, ops)
            ops.append(s[i])

        i += 1

    while ops:
        calc(nums, ops)

    return nums[0]


expr = "-1-(1+(-2)+(3+4+(4-3)))"
print(evaluate_expression(expr), eval(expr))
```

```python
# find max in a window of 4
from collections import deque

def max_in_sliding_window(nums):
    if not nums or len(nums) < 4:
        return []

    result = []
    dq = deque()

    for i in range(len(nums)):
        while dq and dq[0] <= i - 4:
            dq.popleft()

        while dq and nums[dq[-1]] < nums[i]:
            dq.pop()

        dq.append(i)

        if i >= 3:
            result.append(nums[dq[0]])

    return result
    
```
