### Time Complexity O(M*K), where M - target, K - coins quantity

```python
memo = {}

def min_ignore_None(a, b):
	if a is None:
		return b
	if b is None:
		return a
	return min(a, b)

def minimum_coin(m, coins):
	if m in memo:
		return memo[m]
	if m == 0:
		return 0
		
	else:
		answer = None
		for coin in coins:
			subproblem = m - coin
			if subproblem < 0:
				#skip solutions where we try to reach [m] from a negative subproblem
				continue 
			answer = min_ignore_None(answer, minimum_coin(subproblem, coins)+1)
	memo[m] = answer
	return answer
	
```

```python
print(minimum_coin(28, [1, 4, 5]))
```

### Iterative

```python
def minimum_coin_tab(m, coins):
    memo = {0: 0}
    
    for i in range(1,  m+1):
        for coin in coins:
            subproblem = i - coin
            if subproblem < 0:
                continue
            memo[i] = min_ignore_None(memo.get(i), memo[subproblem] + 1)
    return memo.get(m)
```

```python
print(minimum_coin(28, [1, 4, 5]))
```
