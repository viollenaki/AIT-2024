```python
import numpy as np
from collections import Counter

def knn_numpy(X_train, y_train, new_point, k=3):
    # Convert to NumPy arrays
    X_train = np.array(X_train)
    y_train = np.array(y_train)
    new_point = np.array(new_point)

    # Compute Euclidean distances (vectorized)
    distances = np.sqrt(np.sum((X_train - new_point) ** 2, axis=1))

    # Get indices of k nearest neighbors
    k_indices = np.argsort(distances)[:k]

    # Get labels of k nearest neighbors
    k_labels = y_train[k_indices]

    # Majority vote
    most_common = Counter(k_labels).most_common(1)[0][0]

    return most_common


# Example dataset
X_train = np.array([
    [1, 2],
    [2, 3],
    [3, 3],
    [6, 5],
    [7, 7]
])

y_train = np.array(["A", "A", "A", "B", "B"])

new_point = np.array([3, 2])

prediction = knn_numpy(X_train, y_train, new_point, k=3)
print("Predicted class:", prediction)
```
