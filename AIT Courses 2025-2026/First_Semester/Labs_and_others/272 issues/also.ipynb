```python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    
    return merge(left, right)

def merge(left, right):
    result = []
    i = j = 0
    
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    
    result.extend(left[i:])
    result.extend(right[j:])
    return result
```

```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        if not swapped:
            break
    return arr
```

```python
def bucket_sort(arr):
    if len(arr) == 0:
        return arr
    
    num_buckets = 10
    buckets = [[] for _ in range(num_buckets)]
    
    for x in arr:
        index = int(x * num_buckets)
        if index == num_buckets:
            index -= 1
        buckets[index].append(x)
    
    for i in range(num_buckets):
        buckets[i].sort()
    
    result = []
    for bucket in buckets:
        result.extend(bucket)
    return result   
```

```python
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        
        arr[j + 1] = key
    return arr
```

```python
def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    
    return quick_sort(left) + middle + quick_sort(right)
```

```python
697
class Solution:
    def findShortestSubArray(self, nums):
        first = {}   # first index where each number appears
        last = {}    # last index where each number appears
        count = {}   # frequency of each number

        for i, x in enumerate(nums):
            if x not in first:
                first[x] = i
            last[x] = i
            count[x] = count.get(x, 0) + 1

        degree = max(count.values())
        ans = len(nums)

        # For every number with max frequency, compute subarray length
        for x in count:
            if count[x] == degree:
                ans = min(ans, last[x] - first[x] + 1)

        return ans
```

```python
748
class Solution:
    def shortestCompletingWord(self, licensePlate, words):
        # Count required letters from licensePlate
        req = {}
        for c in licensePlate.lower():
            if c.isalpha():
                req[c] = req.get(c, 0) + 1

        def is_completing(word):
            count = {}
            for c in word.lower():
                count[c] = count.get(c, 0) + 1

            # Check if word covers all required characters
            for ch in req:
                if count.get(ch, 0) < req[ch]:
                    return False
            return True

        best = None

        # Try each word and choose the shortest valid one
        for w in words:
            if is_completing(w):
                if best is None or len(w) < len(best):
                    best = w
        
        return best
```

```python
1275
class Solution:
    def tictactoe(self, moves):
        board = [[0] * 3 for _ in range(3)]

        # Заполняем поле
        for i, (r, c) in enumerate(moves):
            board[r][c] = 1 if i % 2 == 0 else -1  # A = 1, B = -1

        # Проверка строк и столбцов
        for i in range(3):
            if abs(sum(board[i])) == 3:
                return "A" if sum(board[i]) == 3 else "B"
            if abs(board[0][i] + board[1][i] + board[2][i]) == 3:
                return "A" if (board[0][i] + board[1][i] + board[2][i]) == 3 else "B"

        # Проверка диагоналей
        diag1 = board[0][0] + board[1][1] + board[2][2]
        diag2 = board[0][2] + board[1][1] + board[2][0]

        if abs(diag1) == 3:
            return "A" if diag1 == 3 else "B"
        if abs(diag2) == 3:
            return "A" if diag2 == 3 else "B"

        # Если все ходы сделаны
        if len(moves) == 9:
            return "Draw"

        return "Pending"
```

```python
1370
class Solution:
    def sortString(self, s: str) -> str:
        # Счётчик букв
        d = {}
        for c in s:
            d[c] = d.get(c, 0) + 1

        outp = []

        while sum(d.values()) > 0:
            # По возрастанию
            for c in sorted(d.keys()):
                if d[c] > 0:
                    outp.append(c)
                    d[c] -= 1

            # По убыванию
            for c in sorted(d.keys(), reverse=True):
                if d[c] > 0:
                    outp.append(c)
                    d[c] -= 1

        return ''.join(outp)

```
