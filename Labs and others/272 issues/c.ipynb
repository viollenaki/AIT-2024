```python
class Solution:
    def buddyStrings(self, s, goal):
        if len(s) != len(goal):
            return False
        
        if s == goal:
            # Нужно хотя бы две одинаковые буквы, чтобы сделать swap без изменения строки
            return len(set(s)) < len(s)
        
        # Находим все позиции, где символы различаются
        diff = [(a, b) for a, b in zip(s, goal) if a != b]
        
        # Если различий ровно 2 и они зеркальны → можно поменять
        print(diff)
        print(diff[0], diff[1][::-1])
        return len(diff) == 2 and diff[0] == diff[1][::-1]

# Пример использования
sol = Solution()
print(sol.buddyStrings("ab", "ba"))  # True
```

```python
class Solution:
    def isAlienSorted(self, words, order):
        # Создаем словарь с позицией каждой буквы в алфавите пришельцев
        rank = {c: i for i, c in enumerate(order)}
        
        def in_order(w1, w2):
            # Сравниваем буквы двух слов по алфавиту пришельцев
            for c1, c2 in zip(w1, w2):
                if rank[c1] < rank[c2]:
                    return True
                elif rank[c1] > rank[c2]:
                    return False
            # Если первые min(len(w1), len(w2)) букв совпали, то короче слово идет раньше
            return len(w1) <= len(w2)
        
        # Проверяем все соседние пары слов
        return all(in_order(words[i], words[i+1]) for i in range(len(words)-1))

# Пример использования
sol = Solution()
print(sol.isAlienSorted(["hello","leetcode"], "lhabcdefgijkmnopqrstuvwxyz"))  #
```

```python
from typing import Optional


class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def mergeTrees(self, root1: Optional[TreeNode], root2: Optional[TreeNode]) -> Optional[TreeNode]:
        if not root1:
            return root2
        if not root2:
            return root1
        
        # Merge the values
        merged = TreeNode(root1.val + root2.val)
        # Recursively merge left and right children
        merged.left = self.mergeTrees(root1.left, root2.left)
        merged.right = self.mergeTrees(root1.right, root2.right)
        
        return merged

# Пример использования
sol = Solution()
n1 = TreeNode(1)
n1.left = TreeNode(3)
n1.right = TreeNode(2)
n1.left.left = TreeNode(5)

n2 = TreeNode(2)
n2.left = TreeNode(1)
n2.right = TreeNode(3)
n2.left.right = TreeNode(4)
n2.right.right = TreeNode(7)

merged_tree = sol.mergeTrees(n1, n2)
vals = []
def inorder_traversal(node):
    if node:
        vals.append(node.val)
        inorder_traversal(node.left)
        inorder_traversal(node.right)
inorder_traversal(merged_tree)  # 617
print(vals)  # [5, 4, 4, 3, 5, 7]
```

```python
class Solution:
    def maxSubsequence(self, nums: List[int], k: int) -> List[int]:
        indexed = [(num, i) for i, num in enumerate(nums)]
        top_k = sorted(indexed, key=lambda x: x[0], reverse=True)[:k]
        top_k_sorted = sorted(top_k, key=lambda x : x[1])
        return [num for num, i in top_k_sorted]
```

```python
from typing import List

class Solution:
    def mergeSimilarItems(self, items1: List[List[int]], items2: List[List[int]]) -> List[List[int]]:
        weights = {}

        # суммируем веса из items1
        for value, weight in items1:
            weights[value] = weights.get(value, 0) + weight

        # суммируем веса из items2
        for value, weight in items2:
            weights[value] = weights.get(value, 0) + weight

        # формируем результат, отсортированный по value
        return [[value, weights[value]] for value in sorted(weights)]

s = Solution()
print(s.mergeSimilarItems([[1,1],[4,5],[3,8]], [[3,1],[1,5]]))
```

```python
class LinkedList:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def merge(h1, h2):
    arr = []
    def unwrap(h):
        curr = h
        prev = None
        while curr:
            arr.append(curr.val)
            prev = curr
            curr = curr.next
    unwrap(h1)
    unwrap(h2)
    arr.sort(reverse=True)
    newl = LinkedList(arr[0])
    for i in arr[1:]:
        newl = LinkedList(i, newl)
    return newl

def print_list(head):
    curr = head
    while curr:
        print(curr.val, end=" -> ")
        curr = curr.next
    print("None")

l1 = LinkedList(1, LinkedList(4, LinkedList(5)))
l2 = LinkedList(2, LinkedList(3, LinkedList(6)))

merged = merge(l1, l2)

print_list(merged)

```

```python
import base64
from collections import Counter, defaultdict
import string

encrypted_text = "KhkTAhMHNFpdQRRZGxBNBBYhXxASFEtVVE1BTmASEBIUS1VUTUFOYBIQEhRLVVRNQU5gEhASFEtVVE1BTmASEBIUWkFaXVBAcgICBD5hOxUABFRgbW9tazQqKzI+MR9tb21rNCorMj4xH21vbWs0KisyPjEfbW9tazQqKzI+MR9tOgMdSzMdAwVOKVQQYlUfHVQoGQczRkMSXQVVMx8AHigSGH5RDgEXAgULYAMJBQVCfy0CFE4hQFUSUwIDEQNbZC4SXl1QDgZURVFONF0QXBlaXH4sD04hQEJTTUsQEAoEHWBFWFdGDlURCQYLM2lZbxRWVS8MCEJgUFlvFAYQFQMSTjRaVUBRTAZUDA9OJVZXVxQJEAAaBAsuElFbFAobEE0DB0pmR10UBRoQCBJUYEFfR0YIEFQMDwpgVlVBQAIbFRkIAS44YldAHgcaTRUcNVcQW1JLARwIEwtgW0MSVUsDFQEICmBCUUZcSxMGAgxOM11FQFcOVQACQQolQURbWgoBHQIPQmBdRFpRGQIdHgROJlNcQVFFfyMFABpgW0MSQAMQVBkIAyUSUVxQSwYEDAILYFFfX0QHEAwEFRdgXVYSTQQABk0SASxHRFtbBUpUZyQWIV9AXlFRfz0DERs0CDpcFFZVR2cECidXQxIJSy4vXU1fHR5rAxhZKFg2U0Jwb204RwQABg4ETn0SADhQDgYABA8PNFtfXBRWVUZnLhs0QkVGDksBBhgEZAVKQF5VBRQABA4AejgQAhSJ8+ZNUE6itKISBkscB00ATjZTXFtQSwUVGQlASjgCGxQmFAwEDBstEnRXRB8dVAIHTgJbXlNGElUgHwQLYBp8V1EfFhsJBE5xAgQbPiwcAggPTjRaVRJGBBoATQ4IYFMQUF0FFAYUQRoyV1UeFBkQABgTAGBbREEUBhQMBAwbLRJUV0QfHVpnJQswRlgSCUsbAQADCzISX1QUBRoQCBJOJkBfXxQZGhsZQRovEkRaUUsTFR8VBiVBRBJYDhQSQ2s5KFNEEl0YVQAFBE40W11XFAobEE0SHiFRVRJXBBgEAQQWKUZJElsNVQ0CFBxgQV9eQR8cGwNeTkp3SFNZGxkRV2snLkJFRg5hVVRNQU5gAToSFEtVVEJBTmASEAsUS0dEZ0FOYBIQEhREVVRNQU5gEgEHFEtCfiIUGjBHRAgUWH9HREE9IV9VEmAZEBFNSSIlV0RRWw8QVFxRXmk4d1tCDhtUGRYBYFBZXFUZDFQZEwslQRBCFAobEE0QQmBRWFdXAFUdC0EaKFdJElUZEFQEBQsuRllRVQdPfj4AAyUSQ0ZGHhYAGBMLSmFRX1FLAxUBFAszOGJXQB4HGk0VHDVXEF1GSxMVARILbjhnWlUfVR0eQRooVxBGXQYQVAwPCmBBQFNXDlUXAgweLFdIW0ASVRsLQRcvR0ISRwQZARkIAS4NEDg+X1xUPg4cNBJxQEYKDFQPGE4JXFNAUQoGHQMGTgZAVUNBDhsXFEFGDHEQAw1YQ11nJgc2V14SVQVVFR8TDzkSXkdZGFlUHg4cNBJZRg5hNw1NCAAjQFVTRwIbE00HHCVDRVdaCAx+JAdOJkBVQ0EOGxcEBB1gU0JXFA4EAQwNQmBBX0BASxcNTQULI0BVU0cCGxNNFw8sR1U4bQQAVA4AAC5dRBJBGBBUCBkHM0ZZXFNLBhsfFQcuVRBTWAwaBgQVBi1BHBJDGRwACEEXL0dCElscG1pNQWQXWlFGFAIGVBkJC2BGWV9RSxQaCUEdMFNTVxQIGhkdDQs4W0RLFAQTVBQOGzISQ11YHgEdAg9RYDh1SlUGBRgIW2QJXEBHQFFVGhgMHWAPEGkFR0RYX01cbAAcAWlhOgEZERs0CBASb1hZRUFQQnIeAh4GNn9+"

# First, decode from base64
encrypted_bytes = base64.b64decode(encrypted_text)
print(f"Encrypted data length: {len(encrypted_bytes)} bytes")

# 1. Look for repeating patterns that might indicate key length
print("\n" + "="*80)
print("1. ANALYZING FOR REPEATING PATTERNS (Key Length Detection)")
print("="*80)

# Method 1: Index of Coincidence (IC) for different key lengths
def index_of_coincidence(data, key_len):
    """Calculate index of coincidence for a given key length"""
    slices = []
    for i in range(key_len):
        slice_data = data[i::key_len]
        if len(slice_data) < 2:
            return 0
        freq = Counter(slice_data)
        total = len(slice_data)
        ic = sum(count * (count - 1) for count in freq.values()) / (total * (total - 1))
        slices.append(ic)
    return sum(slices) / len(slices)

# Try different key lengths
print("\nIndex of Coincidence for different key lengths:")
for key_len in range(1, 21):
    ic = index_of_coincidence(encrypted_bytes, key_len)
    print(f"Key length {key_len:2d}: IC = {ic:.6f}")

# Method 2: Hamming distance between blocks
print("\n\nNormalized Hamming distance for different key lengths:")
for key_len in range(1, 21):
    # Take first few blocks
    num_blocks = min(10, len(encrypted_bytes) // key_len)
    total_dist = 0
    count = 0
    
    for i in range(num_blocks - 1):
        block1 = encrypted_bytes[i*key_len:(i+1)*key_len]
        block2 = encrypted_bytes[(i+1)*key_len:(i+2)*key_len]
        if len(block1) == len(block2):
            # Calculate Hamming distance
            dist = sum(bin(a ^ b).count('1') for a, b in zip(block1, block2))
            total_dist += dist / key_len
            count += 1
    
    if count > 0:
        avg_dist = total_dist / count
        print(f"Key length {key_len:2d}: Avg Hamming distance = {avg_dist:.4f}")

# 2. Look at byte frequencies
print("\n\n" + "="*80)
print("2. BYTE FREQUENCY ANALYSIS")
print("="*80)

byte_counts = Counter(encrypted_bytes)
print(f"Most common bytes in encrypted data:")
for byte, count in byte_counts.most_common(20):
    print(f"  Byte 0x{byte:02x} ({chr(byte) if 32 <= byte < 127 else '·'}): {count:4d} times ({count/len(encrypted_bytes)*100:.2f}%)")

# 3. Try to find the key by assuming the plaintext contains spaces (0x20)
# Space is the most common character in English text
print("\n\n" + "="*80)
print("3. FINDING KEY BY ASSUMING SPACES IN PLAINTEXT")
print("="*80)

# Try different key lengths from our analysis
for key_len in [4, 8, 12, 16]:  # Common key lengths
    print(f"\nTrying key length {key_len}:")
    
    # For each position in the key
    key_guess = [None] * key_len
    for pos in range(key_len):
        # Get all bytes at this position in the key
        slice_bytes = encrypted_bytes[pos::key_len]
        
        # If we assume the most common byte in this slice corresponds to space (0x20)
        # Then key_byte = encrypted_byte ^ 0x20
        if slice_bytes:
            # Try the most common byte
            most_common_byte = Counter(slice_bytes).most_common(1)[0][0]
            key_byte = most_common_byte ^ 0x20
            
            # Check if it's printable
            if 32 <= key_byte < 127:
                key_guess[pos] = chr(key_byte)
                print(f"  Position {pos}: most likely key char = '{chr(key_byte)}' (0x{key_byte:02x})")
            else:
                print(f"  Position {pos}: non-printable key byte 0x{key_byte:02x}")

# 4. Look for common plaintext patterns
print("\n\n" + "="*80)
print("4. LOOKING FOR COMMON PLAINTEXT PATTERNS")
print("="*80)

# Common patterns in exam questions/code
common_patterns = [
    b"def ", b"class ", b"import ", b"return ", b"if ", b"for ", b"while ",
    b"# ", b"\"\"\"", b"'''", b"Question", b"Problem", b"Implement",
    b"function", b"algorithm", b"python", b"leetcode", b"test", b"exam"
]

# Try XORing with common patterns at different positions
print("\nTrying to find key fragments by XORing with common patterns:")
for pattern in common_patterns[:10]:  # First 10 patterns
    pattern_len = len(pattern)
    print(f"\nPattern: {pattern} ({pattern_len} bytes)")
    
    # Slide through the encrypted data
    for i in range(min(100, len(encrypted_bytes) - pattern_len)):
        # XOR the encrypted bytes with the pattern
        possible_key = bytes([encrypted_bytes[i + j] ^ pattern[j] for j in range(pattern_len)])
        
        # Check if the key fragment looks reasonable (printable ASCII)
        if all(32 <= b < 127 for b in possible_key):
            # Check if it looks like a word/phrase
            key_str = possible_key.decode('ascii')
            if any(c.isalpha() for c in key_str):
                print(f"  At position {i}: possible key fragment = '{key_str}'")

# 5. Try to find repeating XOR key using Kasiski examination
print("\n\n" + "="*80)
print("5. KASISKI EXAMINATION (Finding repeating sequences)")
print("="*80)

# Find repeating sequences of 3+ bytes
sequence_length = 3
positions = defaultdict(list)

for i in range(len(encrypted_bytes) - sequence_length + 1):
    sequence = bytes(encrypted_bytes[i:i+sequence_length])
    positions[sequence].append(i)

# Find sequences that appear multiple times
repeating_seqs = {seq: pos_list for seq, pos_list in positions.items() if len(pos_list) > 1}

print(f"Found {len(repeating_seqs)} repeating sequences of length {sequence_length}")
if repeating_seqs:
    print("Most frequently repeating sequences:")
    for seq, pos_list in sorted(repeating_seqs.items(), key=lambda x: len(x[1]), reverse=True)[:10]:
        seq_str = ''.join(chr(b) if 32 <= b < 127 else '.' for b in seq)
        print(f"  Sequence {seq.hex()} ({seq_str}): positions {pos_list}")
        
        # Calculate distances between occurrences
        if len(pos_list) > 1:
            distances = [pos_list[i+1] - pos_list[i] for i in range(len(pos_list)-1)]
            print(f"    Distances: {distances}")
            
            # The key length is likely a divisor of these distances
            print(f"    Possible key lengths (divisors of distances): ", end="")
            for d in distances[:3]:  # First few distances
                for div in range(2, min(20, d//2 + 1)):
                    if d % div == 0:
                        print(f"{div} ", end="")
            print()

# 6. Try common short keys (1-4 chars) that might be used
print("\n\n" + "="*80)
print("6. BRUTE FORCE COMMON SHORT KEYS")
print("="*80)

common_short_keys = [
    "key", "secret", "code", "test", "quiz", "pass", "word", "cipher",
    "xor", "crypt", "hide", "lock", "safe", "data", "file", "text",
    "doc", "exam", "test", "123", "abc", "xyz", "qwe", "asd",
    "key1", "pass1", "test1", "quiz1", "code1", "secret1"
]

for key in common_short_keys:
    try:
        # Simple XOR decryption
        key_bytes = key.encode('utf-8')
        decrypted = bytearray()
        for i in range(min(100, len(encrypted_bytes))):  # Only first 100 bytes
            decrypted.append(encrypted_bytes[i] ^ key_bytes[i % len(key_bytes)])
        
        # Check if it looks like text/code
        decrypted_str = decrypted.decode('utf-8', errors='ignore')
        
        # Check for common patterns in decrypted text
        score = 0
        if any(c.isalpha() for c in decrypted_str[:50]):
            score += 1
        if ' ' in decrypted_str[:50]:
            score += 1
        if any(pattern in decrypted_str.lower() for pattern in ['def', 'class', 'import', 'return']):
            score += 2
        if any(word in decrypted_str.lower() for word in ['path', 'exists', 'sort', 'frequency']):
            score += 3
            
        if score >= 3:  # Good match
            print(f"\nPotential key: '{key}' (score: {score})")
            print(f"First 100 chars decrypted: {decrypted_str[:100]}")
            
    except:
        continue

# 7. Try to guess based on byte distributions
print("\n\n" + "="*80)
print("7. ANALYZING BYTE DISTRIBUTIONS FOR EACH KEY POSITION")
print("="*80)

# Based on earlier analysis, let's try key length 4 (seems likely from IC)
key_len = 4
print(f"\nAnalyzing for key length {key_len}:")

# For each key position, show the byte distribution
for pos in range(key_len):
    slice_bytes = encrypted_bytes[pos::key_len]
    print(f"\nPosition {pos} (total {len(slice_bytes)} bytes):")
    
    # Show most common bytes
    freq = Counter(slice_bytes)
    print("  Most common bytes:")
    for byte, count in freq.most_common(5):
        percent = count / len(slice_bytes) * 100
        char = chr(byte) if 32 <= byte < 127 else '·'
        
        # If we assume this byte is space (0x20), what would the key byte be?
        key_from_space = byte ^ 0x20
        
        # If we assume this byte is 'e' (0x65, most common English letter), what would the key byte be?
        key_from_e = byte ^ 0x65
        
        print(f"    0x{byte:02x} ({char}): {count:3d} ({percent:5.1f}%) -> "
              f"if space: key={chr(key_from_space) if 32 <= key_from_space < 127 else '·'} (0x{key_from_space:02x}), "
              f"if 'e': key={chr(key_from_e) if 32 <= key_from_e < 127 else '·'} (0x{key_from_e:02x})")

# Try to reconstruct key from most common byte as space assumption
print(f"\nKey guess assuming most common byte at each position is space (0x20):")
key_guess = []
for pos in range(key_len):
    slice_bytes = encrypted_bytes[pos::key_len]
    if slice_bytes:
        most_common = Counter(slice_bytes).most_common(1)[0][0]
        key_byte = most_common ^ 0x20
        key_guess.append(key_byte)
        
key_str = ''.join(chr(b) if 32 <= b < 127 else f'\\x{b:02x}' for b in key_guess)
print(f"  Key (hex): {''.join(f'{b:02x}' for b in key_guess)}")
print(f"  Key (ASCII): '{key_str}'")

# Try decrypting with this key
if all(32 <= b < 127 for b in key_guess):
    print(f"\nTrying decryption with key '{key_str}':")
    key_bytes = bytes(key_guess)
    decrypted = bytearray()
    for i in range(min(500, len(encrypted_bytes))):
        decrypted.append(encrypted_bytes[i] ^ key_bytes[i % key_len])
    
    try:
        decrypted_str = decrypted.decode('utf-8')
        print(f"First 200 characters: {decrypted_str[:200]}")
        
        # Check for your keywords
        for word in ['path', 'exists', 'sort', 'frequency']:
            if word in decrypted_str.lower():
                print(f"  Found '{word}' in decrypted text!")
    except:
        print("Failed to decode as UTF-8")
```
