```python
class Solution:
    def buddyStrings(self, s, goal):
        if len(s) != len(goal):
            return False
        
        if s == goal:
            # Нужно хотя бы две одинаковые буквы, чтобы сделать swap без изменения строки
            return len(set(s)) < len(s)
        
        # Находим все позиции, где символы различаются
        diff = [(a, b) for a, b in zip(s, goal) if a != b]
        
        # Если различий ровно 2 и они зеркальны → можно поменять
        print(diff)
        print(diff[0], diff[1][::-1])
        return len(diff) == 2 and diff[0] == diff[1][::-1]

# Пример использования
sol = Solution()
print(sol.buddyStrings("ab", "ba"))  # True
```

```python
from collections import Counter
from math import gcd
from functools import reduce

class Solution:
    def hasGroupsSizeX(self, deck):
        count = Counter(deck).values()
        # Ищем наибольший общий делитель (gcd) всех частот
        print(count)
        print(reduce(gcd, count))
        return reduce(gcd, count) >= 2

# Пример использования
sol = Solution()
print(sol.hasGroupsSizeX([1,2,2,1,1,1, 3,3,3,3]))
```

```python
class Solution:
    def isAlienSorted(self, words, order):
        # Создаем словарь с позицией каждой буквы в алфавите пришельцев
        rank = {c: i for i, c in enumerate(order)}
        
        def in_order(w1, w2):
            # Сравниваем буквы двух слов по алфавиту пришельцев
            for c1, c2 in zip(w1, w2):
                if rank[c1] < rank[c2]:
                    return True
                elif rank[c1] > rank[c2]:
                    return False
            # Если первые min(len(w1), len(w2)) букв совпали, то короче слово идет раньше
            return len(w1) <= len(w2)
        
        # Проверяем все соседние пары слов
        return all(in_order(words[i], words[i+1]) for i in range(len(words)-1))

# Пример использования
sol = Solution()
print(sol.isAlienSorted(["hello","leetcode"], "lhabcdefgijkmnopqrstuvwxyz"))  #
```

```python
from collections import defaultdict, deque

class Solution:
    def validPath(self, n, edges, source, destination):
        # Строим граф в виде списка смежности
        graph = defaultdict(list)
        for u, v in edges:
            graph[u].append(v)
            graph[v].append(u)
        print(graph)
        
        visited = set()
        queue = deque([source])
        print(queue)
        
        while queue:
            node = queue.popleft()
            if node == destination:
                return True
            if node in visited:
                continue
            visited.add(node)
            for neighbor in graph[node]:
                if neighbor not in visited:
                    queue.append(neighbor)
        
        return False
# Пример использования
sol = Solution()
print(sol.validPath(3, [[0,1],[1,2],[2,0]], 0, 2))  # True
```

```python
from typing import Optional


class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def mergeTrees(self, root1: Optional[TreeNode], root2: Optional[TreeNode]) -> Optional[TreeNode]:
        if not root1:
            return root2
        if not root2:
            return root1
        
        # Merge the values
        merged = TreeNode(root1.val + root2.val)
        # Recursively merge left and right children
        merged.left = self.mergeTrees(root1.left, root2.left)
        merged.right = self.mergeTrees(root1.right, root2.right)
        
        return merged

# Пример использования
sol = Solution()
n1 = TreeNode(1)
n1.left = TreeNode(3)
n1.right = TreeNode(2)
n1.left.left = TreeNode(5)

n2 = TreeNode(2)
n2.left = TreeNode(1)
n2.right = TreeNode(3)
n2.left.right = TreeNode(4)
n2.right.right = TreeNode(7)

merged_tree = sol.mergeTrees(n1, n2)
vals = []
def inorder_traversal(node):
    if node:
        vals.append(node.val)
        inorder_traversal(node.left)
        inorder_traversal(node.right)
inorder_traversal(merged_tree)  # 617
print(vals)  # [5, 4, 4, 3, 5, 7]
```
