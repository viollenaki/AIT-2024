```python
def solve_sudoku(board):
    empty = find_empty_cell(board)
    if not empty:
        return True  # solved
    
    r, c = empty
    
    for num in range(1, 10):
        if is_valid(board, r, c, num):
            board[r][c] = num
            
            if solve_sudoku(board):
                return True
            
            board[r][c] = 0  # backtrack
    
    return False  # no valid number found


def find_empty_cell(board):
    for r in range(9):
        for c in range(9):
            if board[r][c] == 0:
                return r, c
    return None


def is_valid(board, r, c, num):
    # row check
    if any(board[r][i] == num for i in range(9)):
        return False
    
    # column check
    if any(board[i][c] == num for i in range(9)):
        return False
    
    # 3×3 box check
    br = (r // 3) * 3
    bc = (c // 3) * 3
    
    for i in range(br, br + 3):
        for j in range(bc, bc + 3):
            if board[i][j] == num:
                return False
    
    return True
```

```python
board = [
    [5,3,0, 0,7,0, 0,0,0],
    [6,0,0, 1,9,5, 0,0,0],
    [0,9,8, 0,0,0, 0,6,0],
    
    [8,0,0, 0,6,0, 0,0,3],
    [4,0,0, 8,0,3, 0,0,1],
    [7,0,0, 0,2,0, 0,0,6],
    
    [0,6,0, 0,0,0, 2,8,0],
    [0,0,0, 4,1,9, 0,0,5],
    [0,0,0, 0,8,0, 0,7,9]
]

if solve_sudoku(board):
    for row in board:
        print(row)
else:
    print("No solution exists")
```
